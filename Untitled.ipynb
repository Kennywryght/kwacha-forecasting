```python
import sys
import os
import joblib

# Add backend to Python path (so that 'ml' is importable)
sys.path.insert(0, os.path.join(os.getcwd(), "backend"))

# Now load the model
model_path = os.path.join("backend", "ml", "artifacts", "arimax.pkl")
model = joblib.load(model_path)

print("Model loaded:", type(model).__name__)
```

```python
import pandas as pd

# Load the processed dataset
df = pd.read_csv("data/processed/mwk_usd_clean.csv")
df["date"] = pd.to_datetime(df["date"])

# Use the last 30 rows for a quick check
recent_data = df.tail(30).copy()
pred = model.predict(recent_data)

print("Predicted values:", pred["y_pred"])
```

```python
import joblib

# Load the tuned Prophet model
prophet_model = joblib.load("backend/ml/artifacts/prophet.pkl")

# Create future dates (next 30 days)
future = prophet_model.make_future_dataframe(periods=30, freq='D')
forecast = prophet_model.predict(future)

# View the forecasts
print(forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail(10))
```
